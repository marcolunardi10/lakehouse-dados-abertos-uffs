# Importaçao das bibliotecas

```python
import boto3
import pandas as pd
import polars as pl
```

# Iniciar sessão AWS

```python
session = boto3.Session(profile_name='default')
s3 = session.client('s3')
```

# Leitura do arquivo

```python
bucket = 'arquivos-uffs'
key = 'Raw/contratos-uffs/contratos-uffs.csv'

obj = s3.get_object(Bucket=bucket, Key=key)

df = pd.read_csv(obj['Body'])

print(df.head())
```
