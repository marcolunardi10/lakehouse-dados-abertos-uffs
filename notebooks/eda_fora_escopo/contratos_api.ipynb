### 1. Import das bibliotecas e definição da função de leitura

```python
import requests
import pandas as pd

def read_ckan_sql(resource_id):
    url = "https://dados.uffs.edu.br/api/3/action/datastore_search_sql"

    limit = 1000
    offset = 0
    all_records = []

    while True:
        sql = f'''
        SELECT *
        FROM "{resource_id}"
        LIMIT {limit}
        OFFSET {offset}
        '''

        response = requests.get(url, params={"sql": sql})
        data = response.json()

        records = data["result"]["records"]

        if not records:
            break

        all_records.extend(records)
        offset += limit

        print(f"Baixados: {offset}")

    df = pd.DataFrame(all_records)
    return df
```

### 2. Passa o id dos dados pra função de leitura, e transforma o resultado em um dataframe pandas

### Funções úteis

```python
# 1. Função para formatar o número no padrão moeda do Brasil
from pandas.io.formats.style import Styler
def formatar_moeda_br(valor):
    if pd.isna(valor):
        return "R$ 0,00"
    
    # Formata com 2 casas decimais no padrão americano (ex: 1,234,567.89)
    valor_str = f"{valor:,.2f}"
    
    # Troca a vírgula por um 'X' temporário, o ponto por vírgula, e o 'X' por ponto
    valor_br = valor_str.replace(",", "X").replace(".", ",").replace("X", ".")
    
    return f"R$ {valor_br}"
```

### Top 10 maiores contratos firmados

```python
# Converte as colunas financeiras para float
cols_financeiras = ['valor_original', 'valor_atual', 'valor_acumulado', 'saldo']
for col in cols_financeiras:
    df_contratos[col] = pd.to_numeric(df_contratos[col], errors='coerce')

# Filtra as colunas mais importantes e ordena pelos maiores valores acumulados
top_contratos = df_contratos[['numero', 'ano', 'nome_fornecedor', 'objeto', 'tipo_contrato', 'valor_acumulado', 'situacao']]
top_contratos = top_contratos.sort_values(by='valor_acumulado', ascending=False)

top_10_contratos = top_contratos.head(10)

# 3. Aplicando a formatação visual usando o Styler
top_10_contratos.style.format({
    'valor_acumulado': formatar_moeda_br
})
```

```python
# 1. Garante que as colunas são numéricas
df_contratos['valor_atual'] = pd.to_numeric(df_contratos['valor_atual'], errors='coerce')
df_contratos['valor_original'] = pd.to_numeric(df_contratos['valor_original'], errors='coerce')

# 2. Cria a coluna de Custo com Aditivos (Diferença)
df_contratos['custo_aditivos'] = df_contratos['valor_atual'] - df_contratos['valor_original']

# 3. Filtra apenas contratos que tiveram aumento (aditivo > 0)
df_com_aditivo = df_contratos[df_contratos['custo_aditivos'] > 0].copy()

# 4. Seleciona as colunas de interesse para a análise
colunas_analise = [
    'numero', 'ano', 'nome_fornecedor', 'objeto', 
    'parcelamento_contrato', 'valor_original', 'valor_atual', 'custo_aditivos'
]
df_com_aditivo = df_com_aditivo[colunas_analise]

# 5. Ordena para ver os maiores "estouros" de orçamento primeiro
df_com_aditivo = df_com_aditivo.sort_values(by='custo_aditivos', ascending=False)

# 6. Exibe os 10 contratos com maiores aditivos, formatando como moeda
df_com_aditivo.head(10).style.format({
    'valor_original': formatar_moeda_br,
    'valor_atual': formatar_moeda_br,
    'custo_aditivos': formatar_moeda_br
})
```
